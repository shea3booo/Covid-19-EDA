general_function_for_comulative_cases

```python
def compute_cumulative_cases(country):
    plt.figure(figsize=(12,6))
    sns.lineplot(x='date_reported', y='cumulative_cases',data= df[df.country == country])
    plt.title(f'Cumulative cases per time in {country}')
```

Genaral function for cumulative deathes

```python
def cumulative_deaths(country):
    plt.figure(figsize=(12,6))
    sns.lineplot(x = 'date_reported', y = 'cumulative_deaths', data = df[df.country == country],marker = 'o')
    plt.title(f'Deaths per time in {country}')
    plt.show()
```

```python
cumulative_deaths('China')
```
